```sql
-- ============================================================
-- UNITY CATALOG SCHEMA SETUP
-- Clinical Trial Intelligence Platform
-- ============================================================
-- Purpose:
-- Create the logical schemas required by the medallion
-- architecture and the data-quality quarantine layer.
--
-- Safe to rerun because IF NOT EXISTS is used.
-- ============================================================


-- ============================================================
-- BRONZE
-- Raw ingested source data
-- ============================================================

CREATE SCHEMA IF NOT EXISTS clinical_trial_intelligence.bronze
COMMENT 'Raw source data ingested from clinical operational systems';


-- ============================================================
-- SILVER
-- Cleaned, standardized and validated data
-- ============================================================

CREATE SCHEMA IF NOT EXISTS clinical_trial_intelligence.silver
COMMENT 'Cleaned, standardized, deduplicated and validated clinical data';


-- ============================================================
-- QUARANTINE
-- Records rejected by Silver data-quality rules
-- ============================================================

CREATE SCHEMA IF NOT EXISTS clinical_trial_intelligence.quarantine
COMMENT 'Records rejected from Silver processing because of data quality or business-rule violations';


-- ============================================================
-- GOLD
-- Business-ready analytical data products
-- ============================================================

CREATE SCHEMA IF NOT EXISTS clinical_trial_intelligence.gold
COMMENT 'Business-ready clinical trial analytical tables and KPI datasets';


-- ============================================================
-- VERIFICATION
-- ============================================================

SHOW SCHEMAS IN clinical_trial_intelligence;




DESCRIBE SCHEMA EXTENDED clinical_trial_intelligence.bronze;
DESCRIBE SCHEMA EXTENDED clinical_trial_intelligence.silver;
DESCRIBE SCHEMA EXTENDED clinical_trial_intelligence.quarantine;
DESCRIBE SCHEMA EXTENDED clinical_trial_intelligence.gold;


ALTER SCHEMA clinical_trial_intelligence.silver
SET MANAGED LOCATION
's3://clinical-trial-intelligence-platform-sk/UnityManaged/silver';


-- ============================================================
-- REMOVE DEVELOPMENT SILVER OUTPUTS
-- ============================================================

DROP SCHEMA clinical_trial_intelligence.quarantine CASCADE;

DROP SCHEMA clinical_trial_intelligence.silver CASCADE;

DROP SCHEMA clinical_trial_intelligence.gold CASCADE;


-- ============================================================
-- RECREATE SILVER WITH S3 MANAGED STORAGE
-- ============================================================

CREATE SCHEMA clinical_trial_intelligence.silver
MANAGED LOCATION
's3://clinical-trial-intelligence-platform-sk/UnityManaged/silver'
COMMENT 'Cleaned, standardized, deduplicated and validated clinical data';


-- ============================================================
-- RECREATE QUARANTINE WITH S3 MANAGED STORAGE
-- ============================================================

CREATE SCHEMA clinical_trial_intelligence.quarantine
MANAGED LOCATION
's3://clinical-trial-intelligence-platform-sk/UnityManaged/quarantine'
COMMENT 'Records rejected from Silver processing because of data quality or business-rule violations';


-- ============================================================
-- RECREATE GOLD WITH S3 MANAGED STORAGE
-- ============================================================

CREATE SCHEMA clinical_trial_intelligence.gold
MANAGED LOCATION
's3://clinical-trial-intelligence-platform-sk/UnityManaged/gold'
COMMENT 'Business-ready clinical trial analytical tables and KPI datasets';
```
