```sql
-- ============================================================
-- SILVER LAYER VALIDATION
-- Clinical Trial Intelligence Platform
-- ============================================================
-- Purpose:
-- Validate the Silver dimension/reference layer and support
-- development and validation of Silver clinical entities.
--
-- Current status:
--   Phase 1 - Dimensions / References : COMPLETE
--   Phase 2 - Subject CDC Investigation: COMPLETE
--   Phase 3 - Subject Processing       : NEXT
-- ============================================================


-- ============================================================
-- SECTION 1: SILVER DATASET INVENTORY
-- ============================================================

SHOW TABLES IN clinical_trial_intelligence.silver;


-- ============================================================
-- SECTION 2: SILVER DIMENSION / REFERENCE ROW COUNTS
-- Expected dimension/reference objects: 13
-- ============================================================

SELECT 'dim_institution' AS table_name, COUNT(*) AS row_count
FROM clinical_trial_intelligence.silver.dim_institution

UNION ALL
SELECT 'dim_sponsor', COUNT(*)
FROM clinical_trial_intelligence.silver.dim_sponsor

UNION ALL
SELECT 'dim_product', COUNT(*)
FROM clinical_trial_intelligence.silver.dim_product

UNION ALL
SELECT 'dim_site', COUNT(*)
FROM clinical_trial_intelligence.silver.dim_site

UNION ALL
SELECT 'dim_study', COUNT(*)
FROM clinical_trial_intelligence.silver.dim_study

UNION ALL
SELECT 'dim_study_arm', COUNT(*)
FROM clinical_trial_intelligence.silver.dim_study_arm

UNION ALL
SELECT 'ref_country_region', COUNT(*)
FROM clinical_trial_intelligence.silver.ref_country_region

UNION ALL
SELECT 'ref_diagnosis', COUNT(*)
FROM clinical_trial_intelligence.silver.ref_diagnosis

UNION ALL
SELECT 'ref_lab_test', COUNT(*)
FROM clinical_trial_intelligence.silver.ref_lab_test

UNION ALL
SELECT 'ref_severity', COUNT(*)
FROM clinical_trial_intelligence.silver.ref_severity

UNION ALL
SELECT 'ref_sex', COUNT(*)
FROM clinical_trial_intelligence.silver.ref_sex

UNION ALL
SELECT 'ref_unit', COUNT(*)
FROM clinical_trial_intelligence.silver.ref_unit

UNION ALL
SELECT 'ref_geography', COUNT(*)
FROM clinical_trial_intelligence.silver.ref_geography

ORDER BY table_name;


-- ============================================================
-- SECTION 3: BUSINESS-KEY COMPLETENESS
-- Expected failed_key_rows = 0 for every object
-- ============================================================

SELECT
    'dim_institution' AS table_name,
    COUNT_IF(institution_id IS NULL OR TRIM(institution_id) = '')
        AS failed_key_rows
FROM clinical_trial_intelligence.silver.dim_institution

UNION ALL
SELECT
    'dim_sponsor',
    COUNT_IF(sponsor_id IS NULL OR TRIM(sponsor_id) = '')
FROM clinical_trial_intelligence.silver.dim_sponsor

UNION ALL
SELECT
    'dim_product',
    COUNT_IF(product_id IS NULL OR TRIM(product_id) = '')
FROM clinical_trial_intelligence.silver.dim_product

UNION ALL
SELECT
    'dim_site',
    COUNT_IF(site_id IS NULL OR TRIM(site_id) = '')
FROM clinical_trial_intelligence.silver.dim_site

UNION ALL
SELECT
    'dim_study',
    COUNT_IF(study_id IS NULL OR TRIM(study_id) = '')
FROM clinical_trial_intelligence.silver.dim_study

UNION ALL
SELECT
    'dim_study_arm',
    COUNT_IF(
        study_id IS NULL OR TRIM(study_id) = ''
        OR arm_code IS NULL OR TRIM(arm_code) = ''
    )
FROM clinical_trial_intelligence.silver.dim_study_arm

UNION ALL
SELECT
    'ref_country_region',
    COUNT_IF(country IS NULL OR TRIM(country) = '')
FROM clinical_trial_intelligence.silver.ref_country_region

UNION ALL
SELECT
    'ref_diagnosis',
    COUNT_IF(diagnosis_code IS NULL OR TRIM(diagnosis_code) = '')
FROM clinical_trial_intelligence.silver.ref_diagnosis

UNION ALL
SELECT
    'ref_lab_test',
    COUNT_IF(lab_test_code IS NULL OR TRIM(lab_test_code) = '')
FROM clinical_trial_intelligence.silver.ref_lab_test

UNION ALL
SELECT
    'ref_severity',
    COUNT_IF(raw_severity IS NULL OR TRIM(raw_severity) = '')
FROM clinical_trial_intelligence.silver.ref_severity

UNION ALL
SELECT
    'ref_sex',
    COUNT_IF(raw_sex IS NULL OR TRIM(raw_sex) = '')
FROM clinical_trial_intelligence.silver.ref_sex

UNION ALL
SELECT
    'ref_unit',
    COUNT_IF(
        lab_test_code IS NULL OR TRIM(lab_test_code) = ''
        OR raw_unit IS NULL OR TRIM(raw_unit) = ''
    )
FROM clinical_trial_intelligence.silver.ref_unit

UNION ALL
SELECT
    'ref_geography',
    COUNT_IF(
        city IS NULL OR TRIM(city) = ''
        OR country IS NULL OR TRIM(country) = ''
    )
FROM clinical_trial_intelligence.silver.ref_geography

ORDER BY table_name;


-- ============================================================
-- SECTION 4: BUSINESS-KEY UNIQUENESS
-- Expected duplicate_groups = 0 for every object
-- ============================================================

SELECT
    'dim_institution' AS table_name,
    COUNT(*) AS duplicate_groups
FROM (
    SELECT institution_id
    FROM clinical_trial_intelligence.silver.dim_institution
    GROUP BY institution_id
    HAVING COUNT(*) > 1
)

UNION ALL
SELECT 'dim_sponsor', COUNT(*)
FROM (
    SELECT sponsor_id
    FROM clinical_trial_intelligence.silver.dim_sponsor
    GROUP BY sponsor_id
    HAVING COUNT(*) > 1
)

UNION ALL
SELECT 'dim_product', COUNT(*)
FROM (
    SELECT product_id
    FROM clinical_trial_intelligence.silver.dim_product
    GROUP BY product_id
    HAVING COUNT(*) > 1
)

UNION ALL
SELECT 'dim_site', COUNT(*)
FROM (
    SELECT site_id
    FROM clinical_trial_intelligence.silver.dim_site
    GROUP BY site_id
    HAVING COUNT(*) > 1
)

UNION ALL
SELECT 'dim_study', COUNT(*)
FROM (
    SELECT study_id
    FROM clinical_trial_intelligence.silver.dim_study
    GROUP BY study_id
    HAVING COUNT(*) > 1
)

UNION ALL
SELECT 'dim_study_arm', COUNT(*)
FROM (
    SELECT study_id, arm_code
    FROM clinical_trial_intelligence.silver.dim_study_arm
    GROUP BY study_id, arm_code
    HAVING COUNT(*) > 1
)

UNION ALL
SELECT 'ref_country_region', COUNT(*)
FROM (
    SELECT LOWER(TRIM(country))
    FROM clinical_trial_intelligence.silver.ref_country_region
    GROUP BY LOWER(TRIM(country))
    HAVING COUNT(*) > 1
)

UNION ALL
SELECT 'ref_diagnosis', COUNT(*)
FROM (
    SELECT diagnosis_code
    FROM clinical_trial_intelligence.silver.ref_diagnosis
    GROUP BY diagnosis_code
    HAVING COUNT(*) > 1
)

UNION ALL
SELECT 'ref_lab_test', COUNT(*)
FROM (
    SELECT lab_test_code
    FROM clinical_trial_intelligence.silver.ref_lab_test
    GROUP BY lab_test_code
    HAVING COUNT(*) > 1
)

UNION ALL
SELECT 'ref_severity', COUNT(*)
FROM (
    SELECT LOWER(TRIM(raw_severity))
    FROM clinical_trial_intelligence.silver.ref_severity
    GROUP BY LOWER(TRIM(raw_severity))
    HAVING COUNT(*) > 1
)

UNION ALL
SELECT 'ref_sex', COUNT(*)
FROM (
    SELECT LOWER(TRIM(raw_sex))
    FROM clinical_trial_intelligence.silver.ref_sex
    GROUP BY LOWER(TRIM(raw_sex))
    HAVING COUNT(*) > 1
)

UNION ALL
SELECT 'ref_unit', COUNT(*)
FROM (
    SELECT
        LOWER(TRIM(lab_test_code)),
        LOWER(TRIM(raw_unit))
    FROM clinical_trial_intelligence.silver.ref_unit
    GROUP BY
        LOWER(TRIM(lab_test_code)),
        LOWER(TRIM(raw_unit))
    HAVING COUNT(*) > 1
)

UNION ALL
SELECT 'ref_geography', COUNT(*)
FROM (
    SELECT
        LOWER(TRIM(city)),
        LOWER(TRIM(country))
    FROM clinical_trial_intelligence.silver.ref_geography
    GROUP BY
        LOWER(TRIM(city)),
        LOWER(TRIM(country))
    HAVING COUNT(*) > 1
)

ORDER BY table_name;


-- ============================================================
-- SECTION 5: REFERENCE VALUE COMPLETENESS
-- Expected missing_mapping_rows = 0
-- ============================================================

SELECT
    'ref_country_region.region' AS validation_rule,
    COUNT_IF(region IS NULL OR TRIM(region) = '')
        AS missing_mapping_rows
FROM clinical_trial_intelligence.silver.ref_country_region

UNION ALL
SELECT
    'ref_diagnosis.diagnosis_description',
    COUNT_IF(
        diagnosis_description IS NULL
        OR TRIM(diagnosis_description) = ''
    )
FROM clinical_trial_intelligence.silver.ref_diagnosis

UNION ALL
SELECT
    'ref_lab_test.standard_unit',
    COUNT_IF(
        standard_unit IS NULL
        OR TRIM(standard_unit) = ''
    )
FROM clinical_trial_intelligence.silver.ref_lab_test

UNION ALL
SELECT
    'ref_severity.standard_severity',
    COUNT_IF(
        standard_severity IS NULL
        OR TRIM(standard_severity) = ''
    )
FROM clinical_trial_intelligence.silver.ref_severity

UNION ALL
SELECT
    'ref_sex.standard_sex',
    COUNT_IF(
        standard_sex IS NULL
        OR TRIM(standard_sex) = ''
    )
FROM clinical_trial_intelligence.silver.ref_sex

UNION ALL
SELECT
    'ref_unit.standard_unit',
    COUNT_IF(
        standard_unit IS NULL
        OR TRIM(standard_unit) = ''
    )
FROM clinical_trial_intelligence.silver.ref_unit

UNION ALL
SELECT
    'ref_geography.region',
    COUNT_IF(
        region IS NULL
        OR TRIM(region) = ''
    )
FROM clinical_trial_intelligence.silver.ref_geography;


-- ============================================================
-- SECTION 6: STUDY ARM ALLOCATION
-- Expected allocation_total = 100 for every study
-- ============================================================

SELECT
    study_id,
    SUM(planned_allocation_pct) AS allocation_total
FROM clinical_trial_intelligence.silver.dim_study_arm
GROUP BY study_id
ORDER BY study_id;


-- ============================================================
-- SECTION 7: STUDY ARM ALLOCATION FAILURES
-- Expected result: ZERO ROWS
-- ============================================================

SELECT
    study_id,
    SUM(planned_allocation_pct) AS allocation_total
FROM clinical_trial_intelligence.silver.dim_study_arm
GROUP BY study_id
HAVING ABS(SUM(planned_allocation_pct) - 100) > 0.0001
ORDER BY study_id;


-- ============================================================
-- SECTION 8: RELATIONSHIP / ORPHAN VALIDATION
-- Expected all counts = 0
-- ============================================================

SELECT
    'PRODUCT_TO_SPONSOR' AS relationship,
    COUNT(*) AS orphan_rows
FROM clinical_trial_intelligence.silver.dim_product p
LEFT JOIN clinical_trial_intelligence.silver.dim_sponsor s
    ON p.sponsor_id = s.sponsor_id
WHERE p.sponsor_id IS NOT NULL
  AND TRIM(p.sponsor_id) <> ''
  AND s.sponsor_id IS NULL

UNION ALL

SELECT
    'STUDY_TO_SPONSOR',
    COUNT(*)
FROM clinical_trial_intelligence.silver.dim_study st
LEFT JOIN clinical_trial_intelligence.silver.dim_sponsor sp
    ON st.sponsor_id = sp.sponsor_id
WHERE st.sponsor_id IS NOT NULL
  AND TRIM(st.sponsor_id) <> ''
  AND sp.sponsor_id IS NULL

UNION ALL

SELECT
    'SITE_TO_STUDY',
    COUNT(*)
FROM clinical_trial_intelligence.silver.dim_site si
LEFT JOIN clinical_trial_intelligence.silver.dim_study st
    ON si.study_id = st.study_id
WHERE si.study_id IS NOT NULL
  AND TRIM(si.study_id) <> ''
  AND st.study_id IS NULL

UNION ALL

SELECT
    'SITE_TO_INSTITUTION',
    COUNT(*)
FROM clinical_trial_intelligence.silver.dim_site si
LEFT JOIN clinical_trial_intelligence.silver.dim_institution i
    ON si.institution_id = i.institution_id
WHERE si.institution_id IS NOT NULL
  AND TRIM(si.institution_id) <> ''
  AND i.institution_id IS NULL

UNION ALL

SELECT
    'ARM_TO_STUDY',
    COUNT(*)
FROM clinical_trial_intelligence.silver.dim_study_arm a
LEFT JOIN clinical_trial_intelligence.silver.dim_study st
    ON a.study_id = st.study_id
WHERE a.study_id IS NOT NULL
  AND TRIM(a.study_id) <> ''
  AND st.study_id IS NULL

UNION ALL

SELECT
    'ARM_TO_PRODUCT',
    COUNT(*)
FROM clinical_trial_intelligence.silver.dim_study_arm a
LEFT JOIN clinical_trial_intelligence.silver.dim_product p
    ON a.product_id = p.product_id
WHERE a.product_id IS NOT NULL
  AND TRIM(a.product_id) <> ''
  AND p.product_id IS NULL;


-- ============================================================
-- SECTION 9: SILVER LINEAGE METADATA COMPLETENESS
-- Expected all missing counts = 0
-- ============================================================

SELECT
    'dim_institution' AS table_name,
    COUNT(*) AS total_rows,
    COUNT_IF(_source_file IS NULL) AS missing_source_file,
    COUNT_IF(_source_file_name IS NULL) AS missing_source_file_name,
    COUNT_IF(_source_file_modification_ts IS NULL)
        AS missing_file_modification_ts,
    COUNT_IF(_ingestion_ts IS NULL) AS missing_ingestion_ts,
    COUNT_IF(_ingestion_date IS NULL) AS missing_ingestion_date
FROM clinical_trial_intelligence.silver.dim_institution

UNION ALL

SELECT
    'dim_sponsor',
    COUNT(*),
    COUNT_IF(_source_file IS NULL),
    COUNT_IF(_source_file_name IS NULL),
    COUNT_IF(_source_file_modification_ts IS NULL),
    COUNT_IF(_ingestion_ts IS NULL),
    COUNT_IF(_ingestion_date IS NULL)
FROM clinical_trial_intelligence.silver.dim_sponsor

UNION ALL

SELECT
    'dim_product',
    COUNT(*),
    COUNT_IF(_source_file IS NULL),
    COUNT_IF(_source_file_name IS NULL),
    COUNT_IF(_source_file_modification_ts IS NULL),
    COUNT_IF(_ingestion_ts IS NULL),
    COUNT_IF(_ingestion_date IS NULL)
FROM clinical_trial_intelligence.silver.dim_product

UNION ALL

SELECT
    'dim_site',
    COUNT(*),
    COUNT_IF(_source_file IS NULL),
    COUNT_IF(_source_file_name IS NULL),
    COUNT_IF(_source_file_modification_ts IS NULL),
    COUNT_IF(_ingestion_ts IS NULL),
    COUNT_IF(_ingestion_date IS NULL)
FROM clinical_trial_intelligence.silver.dim_site

UNION ALL

SELECT
    'dim_study',
    COUNT(*),
    COUNT_IF(_source_file IS NULL),
    COUNT_IF(_source_file_name IS NULL),
    COUNT_IF(_source_file_modification_ts IS NULL),
    COUNT_IF(_ingestion_ts IS NULL),
    COUNT_IF(_ingestion_date IS NULL)
FROM clinical_trial_intelligence.silver.dim_study

UNION ALL

SELECT
    'dim_study_arm',
    COUNT(*),
    COUNT_IF(_source_file IS NULL),
    COUNT_IF(_source_file_name IS NULL),
    COUNT_IF(_source_file_modification_ts IS NULL),
    COUNT_IF(_ingestion_ts IS NULL),
    COUNT_IF(_ingestion_date IS NULL)
FROM clinical_trial_intelligence.silver.dim_study_arm

UNION ALL

SELECT
    'ref_country_region',
    COUNT(*),
    COUNT_IF(_source_file IS NULL),
    COUNT_IF(_source_file_name IS NULL),
    COUNT_IF(_source_file_modification_ts IS NULL),
    COUNT_IF(_ingestion_ts IS NULL),
    COUNT_IF(_ingestion_date IS NULL)
FROM clinical_trial_intelligence.silver.ref_country_region

UNION ALL

SELECT
    'ref_diagnosis',
    COUNT(*),
    COUNT_IF(_source_file IS NULL),
    COUNT_IF(_source_file_name IS NULL),
    COUNT_IF(_source_file_modification_ts IS NULL),
    COUNT_IF(_ingestion_ts IS NULL),
    COUNT_IF(_ingestion_date IS NULL)
FROM clinical_trial_intelligence.silver.ref_diagnosis

UNION ALL

SELECT
    'ref_lab_test',
    COUNT(*),
    COUNT_IF(_source_file IS NULL),
    COUNT_IF(_source_file_name IS NULL),
    COUNT_IF(_source_file_modification_ts IS NULL),
    COUNT_IF(_ingestion_ts IS NULL),
    COUNT_IF(_ingestion_date IS NULL)
FROM clinical_trial_intelligence.silver.ref_lab_test

UNION ALL

SELECT
    'ref_severity',
    COUNT(*),
    COUNT_IF(_source_file IS NULL),
    COUNT_IF(_source_file_name IS NULL),
    COUNT_IF(_source_file_modification_ts IS NULL),
    COUNT_IF(_ingestion_ts IS NULL),
    COUNT_IF(_ingestion_date IS NULL)
FROM clinical_trial_intelligence.silver.ref_severity

UNION ALL

SELECT
    'ref_sex',
    COUNT(*),
    COUNT_IF(_source_file IS NULL),
    COUNT_IF(_source_file_name IS NULL),
    COUNT_IF(_source_file_modification_ts IS NULL),
    COUNT_IF(_ingestion_ts IS NULL),
    COUNT_IF(_ingestion_date IS NULL)
FROM clinical_trial_intelligence.silver.ref_sex

UNION ALL

SELECT
    'ref_unit',
    COUNT(*),
    COUNT_IF(_source_file IS NULL),
    COUNT_IF(_source_file_name IS NULL),
    COUNT_IF(_source_file_modification_ts IS NULL),
    COUNT_IF(_ingestion_ts IS NULL),
    COUNT_IF(_ingestion_date IS NULL)
FROM clinical_trial_intelligence.silver.ref_unit

UNION ALL

SELECT
    'ref_geography',
    COUNT(*),
    COUNT_IF(_source_file IS NULL),
    COUNT_IF(_source_file_name IS NULL),
    COUNT_IF(_source_file_modification_ts IS NULL),
    COUNT_IF(_ingestion_ts IS NULL),
    COUNT_IF(_ingestion_date IS NULL)
FROM clinical_trial_intelligence.silver.ref_geography

ORDER BY table_name;


-- ============================================================
-- SECTION 10: FINAL DIMENSION LAYER GATE
-- Expected result: ZERO ROWS
-- ============================================================

SELECT
    'DIM_SITE_UNMAPPED_REGION' AS failed_check,
    COUNT(*) AS failed_rows
FROM clinical_trial_intelligence.silver.dim_site
WHERE region = 'UNMAPPED'
   OR region IS NULL
HAVING COUNT(*) > 0

UNION ALL

SELECT
    'DIM_SITE_DUPLICATE_KEY',
    COUNT(*)
FROM (
    SELECT site_id
    FROM clinical_trial_intelligence.silver.dim_site
    GROUP BY site_id
    HAVING COUNT(*) > 1
)
HAVING COUNT(*) > 0

UNION ALL

SELECT
    'REF_GEOGRAPHY_DUPLICATE_KEY',
    COUNT(*)
FROM (
    SELECT
        LOWER(TRIM(city)),
        LOWER(TRIM(country))
    FROM clinical_trial_intelligence.silver.ref_geography
    GROUP BY
        LOWER(TRIM(city)),
        LOWER(TRIM(country))
    HAVING COUNT(*) > 1
)
HAVING COUNT(*) > 0;


-- ============================================================
-- SECTION 11: SUBJECT CDC INVESTIGATION
-- DEVELOPMENT / AUDIT SECTION
-- ============================================================
-- This section documents WHY source_snapshot_date is used as
-- the logical sequence for EDC subject snapshots.
-- ============================================================


-- ------------------------------------------------------------
-- 11.1 Bronze Subject Schema
-- ------------------------------------------------------------

DESCRIBE TABLE clinical_trial_intelligence.bronze.edc_subjects;


-- ------------------------------------------------------------
-- 11.2 Sample Bronze Subject Records
-- ------------------------------------------------------------

SELECT *
FROM clinical_trial_intelligence.bronze.edc_subjects
LIMIT 20;


-- ------------------------------------------------------------
-- 11.3 Detect Subjects Appearing in Multiple Snapshot Files
-- ------------------------------------------------------------

SELECT
    subject_id,
    COUNT(*) AS version_count,
    MIN(_ingestion_ts) AS first_ingestion,
    MAX(_ingestion_ts) AS latest_ingestion,
    COUNT(DISTINCT _source_file_name) AS source_files
FROM clinical_trial_intelligence.bronze.edc_subjects
WHERE subject_id IS NOT NULL
  AND TRIM(subject_id) <> ''
GROUP BY subject_id
HAVING COUNT(*) > 1
ORDER BY version_count DESC
LIMIT 20;


-- ------------------------------------------------------------
-- 11.4 CDC Sequence Investigation
-- ------------------------------------------------------------
-- _ingestion_ts cannot be relied upon because multiple source
-- snapshots were ingested during the same Bronze pipeline run.
--
-- _source_file_modification_ts also does not represent the
-- logical business ordering of the snapshots.
--
-- Therefore the YYYYMMDD date encoded in the source filename
-- is used as the source snapshot sequence.
-- ------------------------------------------------------------

SELECT
    subject_id,
    _source_file_name,

    TO_DATE(
        REGEXP_EXTRACT(
            _source_file_name,
            'subjects_(\\d{8})\\.csv',
            1
        ),
        'yyyyMMdd'
    ) AS source_snapshot_date,

    _source_file_modification_ts,
    _ingestion_ts,

    subject_status,
    enrollment_date,
    discontinuation_date,
    arm_code,
    age,
    sex

FROM clinical_trial_intelligence.bronze.edc_subjects

WHERE subject_id IN (
    '104-003-0067',
    '104-010-0065',
    '102-003-0078'
)

ORDER BY
    subject_id,
    source_snapshot_date;


-- ------------------------------------------------------------
-- 11.5 Snapshot Filename Parsing Validation
-- Expected invalid_snapshot_dates = 0
-- ------------------------------------------------------------

SELECT
    COUNT(*) AS invalid_snapshot_dates
FROM clinical_trial_intelligence.bronze.edc_subjects
WHERE TO_DATE(
        REGEXP_EXTRACT(
            _source_file_name,
            'subjects_(\\d{8})\\.csv',
            1
        ),
        'yyyyMMdd'
      ) IS NULL;


-- ------------------------------------------------------------
-- 11.6 SUBJECT CDC SEQUENCE UNIQUENESS CHECK
-- Business key : subject_id
-- Sequence     : source_snapshot_date
--
-- Expected result: ZERO ROWS
-- ------------------------------------------------------------

WITH subject_versions AS (

    SELECT
        TRIM(subject_id) AS subject_id,

        TO_DATE(
            REGEXP_EXTRACT(
                _source_file_name,
                'subjects_(\\d{8})\\.csv',
                1
            ),
            'yyyyMMdd'
        ) AS source_snapshot_date

    FROM clinical_trial_intelligence.bronze.edc_subjects

    WHERE subject_id IS NOT NULL
      AND TRIM(subject_id) <> ''
)

SELECT
    subject_id,
    source_snapshot_date,
    COUNT(*) AS record_count

FROM subject_versions

GROUP BY
    subject_id,
    source_snapshot_date

HAVING COUNT(*) > 1

ORDER BY record_count DESC;


-- ============================================================
-- SECTION 12: SUBJECT PIPELINE DEPENDENCY SCHEMAS
-- Run before implementing/finalizing subjects.py
-- ============================================================

DESCRIBE TABLE clinical_trial_intelligence.silver.ref_sex;

DESCRIBE TABLE clinical_trial_intelligence.silver.ref_diagnosis;

DESCRIBE TABLE clinical_trial_intelligence.silver.dim_study;

DESCRIBE TABLE clinical_trial_intelligence.silver.dim_site;

DESCRIBE TABLE clinical_trial_intelligence.silver.dim_study_arm;


-- ============================================================
-- END OF CURRENT SILVER VALIDATION
-- ============================================================



SELECT
    subject_id,
    _source_file_name,

    TO_DATE(
        REGEXP_EXTRACT(
            _source_file_name,
            'subjects_(\\d{8})\\.csv',
            1
        ),
        'yyyyMMdd'
    ) AS source_snapshot_date,

    _source_file,
    _source_file_modification_ts,
    _ingestion_ts,

    study_id,
    site_id,
    subject_status,
    enrollment_date,
    discontinuation_date,
    arm_code,
    age,
    sex

FROM clinical_trial_intelligence.bronze.edc_subjects

WHERE TRIM(subject_id) = '104-012-0014'

ORDER BY
    source_snapshot_date,
    _source_file_name,
    _source_file;



WITH snapshot_rows AS (
    SELECT
        TRIM(subject_id) AS subject_id,

        TO_DATE(
            REGEXP_EXTRACT(
                _source_file_name,
                'subjects_(\\d{8})\\.csv',
                1
            ),
            'yyyyMMdd'
        ) AS source_snapshot_date,

        _source_file_name,
        _source_file

    FROM clinical_trial_intelligence.bronze.edc_subjects

    WHERE subject_id IS NOT NULL
      AND TRIM(subject_id) <> ''
)

SELECT
    subject_id,
    source_snapshot_date,
    COUNT(*) AS record_count,
    COUNT(DISTINCT _source_file_name) AS file_count,
    COLLECT_SET(_source_file_name) AS source_files

FROM snapshot_rows

GROUP BY
    subject_id,
    source_snapshot_date

HAVING COUNT(*) > 1

ORDER BY
    record_count DESC,
    subject_id;


-- ============================================================
-- SUBJECT SNAPSHOT INVENTORY
-- Purpose:
-- Confirm every source file maps to exactly one snapshot date
-- and inspect number of rows / unique subjects per snapshot.
--
-- Expected:
-- row_count = distinct_subjects for every snapshot
-- ============================================================

SELECT
    _source_file_name,

    TO_DATE(
        REGEXP_EXTRACT(
            _source_file_name,
            'subjects_(\\d{8})\\.csv',
            1
        ),
        'yyyyMMdd'
    ) AS source_snapshot_date,

    COUNT(*) AS row_count,

    COUNT(DISTINCT TRIM(subject_id)) AS distinct_subjects

FROM clinical_trial_intelligence.bronze.edc_subjects

WHERE subject_id IS NOT NULL
  AND TRIM(subject_id) <> ''

GROUP BY
    _source_file_name,

    TO_DATE(
        REGEXP_EXTRACT(
            _source_file_name,
            'subjects_(\\d{8})\\.csv',
            1
        ),
        'yyyyMMdd'
    )

ORDER BY source_snapshot_date;




-- ============================================================
-- REF_SEX NORMALIZED KEY UNIQUENESS CHECK
--
-- Purpose:
-- subjects.py joins using UPPER(TRIM(raw_sex)).
-- Therefore uniqueness must be checked AFTER normalization,
-- not only against the original raw_sex value.
--
-- Expected result: ZERO ROWS
-- ============================================================

SELECT
    UPPER(TRIM(raw_sex)) AS normalized_raw_sex,
    COUNT(*) AS record_count,
    COUNT(DISTINCT UPPER(TRIM(standard_sex))) AS standard_value_count,
    COLLECT_SET(raw_sex) AS raw_values,
    COLLECT_SET(standard_sex) AS standard_values

FROM clinical_trial_intelligence.silver.ref_sex

WHERE raw_sex IS NOT NULL
  AND TRIM(raw_sex) <> ''

GROUP BY
    UPPER(TRIM(raw_sex))

HAVING COUNT(*) > 1

ORDER BY record_count DESC;


-- ============================================================
-- SECTION 15: NORMALIZED REFERENCE KEY VALIDATION
-- ============================================================
--
-- Purpose:
-- Detect reference-table keys that become duplicates after
-- applying the same normalization used by downstream joins.
--
-- This validation identified the ref_sex fan-out condition
-- during development of the Silver subjects pipeline.
--
-- Expected:
-- No normalized key should map to more than one STANDARD value.
--
-- Note:
-- Multiple raw case variants are acceptable when they all map
-- to exactly the same standard value.
-- ============================================================

SELECT
    UPPER(TRIM(raw_sex)) AS normalized_raw_sex,
    COUNT(DISTINCT UPPER(TRIM(standard_sex)))
        AS standard_value_count,
    COLLECT_SET(raw_sex) AS raw_values,
    COLLECT_SET(standard_sex) AS standard_values

FROM clinical_trial_intelligence.silver.ref_sex

WHERE raw_sex IS NOT NULL
  AND TRIM(raw_sex) <> ''

GROUP BY
    UPPER(TRIM(raw_sex))

HAVING COUNT(
    DISTINCT UPPER(TRIM(standard_sex))
) > 1;


-- ============================================================
-- SUBJECTS FINAL SCD TYPE 2 VALIDATION
-- ============================================================


-- ============================================================
-- 1. SUBJECT TABLE BASIC PROFILE
-- ============================================================

SELECT
    COUNT(*) AS total_scd_rows,
    COUNT(DISTINCT subject_id) AS distinct_subjects,
    COUNT_IF(__END_AT IS NULL) AS current_rows
FROM clinical_trial_intelligence.silver.subjects;


-- ============================================================
-- 2. EXACTLY ONE CURRENT RECORD PER SUBJECT
-- Expected result: ZERO ROWS
-- ============================================================

SELECT
    subject_id,
    COUNT(*) AS current_record_count
FROM clinical_trial_intelligence.silver.subjects
WHERE __END_AT IS NULL
GROUP BY subject_id
HAVING COUNT(*) <> 1
ORDER BY current_record_count DESC;


-- ============================================================
-- 3. BUSINESS KEY COMPLETENESS
-- Expected result: 0
-- ============================================================

SELECT
    COUNT(*) AS invalid_subject_keys
FROM clinical_trial_intelligence.silver.subjects
WHERE subject_id IS NULL
   OR TRIM(subject_id) = '';


-- ============================================================
-- 4. SEX STANDARDIZATION
-- Expected:
-- no unexpected values / ideally no UNMAPPED
-- ============================================================

SELECT
    sex,
    COUNT(*) AS record_count
FROM clinical_trial_intelligence.silver.subjects
GROUP BY sex
ORDER BY record_count DESC;


-- ============================================================
-- 5. INSPECT ACTUAL SCD2 HISTORY
-- ============================================================

SELECT
    subject_id,
    subject_status,
    age,
    sex,
    arm_code,
    enrollment_date,
    discontinuation_date,
    source_snapshot_date,
    __START_AT,
    __END_AT
FROM clinical_trial_intelligence.silver.subjects
WHERE subject_id IN (
    '104-003-0067',
    '104-010-0065',
    '102-003-0078'
)
ORDER BY
    subject_id,
    __START_AT;


SELECT *

FROM clinical_trial_intelligence.bronze.edc_visits

LIMIT 20;
```
